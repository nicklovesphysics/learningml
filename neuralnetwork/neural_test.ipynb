```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
import pandas as pd
import neural_nick as nn
import numpy as np
```

```python
bc = load_breast_cancer()       #0 is malignant, 1 is benign.
test_size = .2

input = bc.data
train = bc.target

input_train_no_t, input_test, target_train, target_test = train_test_split(input, train, test_size = test_size, random_state = 20)

input_train = input_train_no_t.T

train_in_shape = np.shape(input_train)[1]
test_in_shape = np.shape(input_test)[1]
print(train_in_shape)
```

```python
loss1 = nn.MeanSquaredError(inputs = input_train[0], layer_number = 1)
activation_1 = nn.ReLU(inputs= input_train, order = 0)
layer_1 = nn.hidden_input_layer(order = 0, n_neurons = train_in_shape, n_inputs = train_in_shape, act_class = activation_1, loss_class = loss1, inputs = input_train)


loss2 = nn.MeanSquaredError(layer_number = 2)  
activation_2 = nn.ReLU(inputs = np.random.randn(32,), order = 1)
layer_2 = nn.hidden_input_layer(order = 1, n_neurons = 10, n_inputs= layer_1.nneur, act_class = activation_2, loss_class = loss2)


loss3 = nn.MeanSquaredError(targets = target_train, layer_number = 3)
activation_3 = nn.ReLU(order = 2)
outputlayer = nn.output_layer(order = 2, n_neurons = 1, n_inputs = layer_2.nneur, act_class = activation_3, loss_class = loss3)



test_activations = [activation_1, activation_2, activation_3]

layers_list = [layer_1, layer_2, outputlayer]

loss_list_test = [loss1, loss2, loss3]


 
gd = nn.GradientDescent()

gd.optimize(10, .01, layers_opt = layers_list, activation_opt = test_activations)     
```
